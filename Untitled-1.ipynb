```python
import numpy as np

class board:
    def __init__(self):
        self.arr = np.zeros((3,3))
        self.turn = True
    def printB(self):
        for i in self.arr:
            print(i)
    def move(self, x, y):
        if(self.arr[x,y] == 0):
            if(self.turn):
                self.arr[x,y] = 1
            else:
                self.arr[x,y] = -1
            self.turn = not self.turn
            return 1
        else:
            return -1
    def checkWin(self, p):
        for y in range(3):
            for x in range(3):
                right = self.winHelper(x, y, 1, 0, p)
                dLeft = self.winHelper(x, y, -1, 1, p)
                dRight = self.winHelper(x, y, 1, 1, p)      
                down = self.winHelper(x, y, 0, 1, p)
                if( right or dLeft or dRight or down):
                    return True
        return False
        
    def winHelper(self, x, y, dx, dy, p):
        if(x + 2 * dx > 2 or x + 2 * dx < 0):
                return False
        if(y + 2 * dy > 2 or y + 2 * dy < 0):
                return False
        if(self.arr[x,y] == p and self.arr[x + dx, y + dy] == p and self.arr[x + 2 * dx, y + 2 *dy] == p):
             return True
        return False


        
        
b = board()
b.move(2, 0)
b.move(0,1)
b.move(1, 0)
b.move(0,2)
b.move(0, 0)
b.move(1, 1)
b.printB()
print(b.winHelper(0, 0, 1, 0, 1))


```
