```python
import numpy as np
class board:
    def __init__(self):
        self.arr = np.zeros((3,3))
        self.turn = True

    def printB(self):
        for i in self.arr:
            print(i)

    def move(self, x, y):
        if(self.arr[x,y] == 0):
            if(self.turn):
                self.arr[x,y] = 1
            else:
                self.arr[x,y] = -1
            self.turn = not self.turn
            return 1
        else:
            return -1
        
    def checkWinTurn(self):
        if(self.turn):
              return self.checkWin(-1)
        return self.checkWin(1)
    
    def checkWin(self, p):
        for y in range(3):
            for x in range(3):
                right = self.winHelper(x, y, 1, 0, p)
                dLeft = self.winHelper(x, y, -1, 1, p)
                dRight = self.winHelper(x, y, 1, 1, p)      
                down = self.winHelper(x, y, 0, 1, p)
                if( right == 1 or dLeft == 1 or dRight == 1 or down == 1):
                    return True
        return False
        
    def winHelper(self, x, y, dx, dy, p):
        if(x + 2 * dx > 2 or x + 2 * dx < 0):
                return False
        if(y + 2 * dy > 2 or y + 2 * dy < 0):
                return False
        if(self.arr[x,y] == p and self.arr[x + dx, y + dy] == p and self.arr[x + 2 * dx, y + 2 *dy] == p):
             return True
        return False
    
    def getBoard(self):
        return self.arr
```

```python
import numpy as np
import math as math
class MSGP:
    def __init__(self, mat, ba):
        self.m = mat
        self.b = ba
        self.layers = len(mat)

    def breakdown(self, board, p):
        arr = np.zeros(9)
        r = 0
        for i in board:
            for k in i:
                arr[r] = k * p
                r += 1
        return arr

    # y = mx + b
    def mxb(self, m, x, b):
        return np.add(b, np.multiply(x, m))

    def move(self, board, p):
        input = self.breakdown(board, p)
        for i in range(self.layers):
            input = self.mxb(self.m[i], input, self.b[i])        

        max = 0
        for i in range(self.layers):
            if(input[max] < input[i]):
                max = i
        
        tmp = np.zeros(2)
        tmp[0] = math.floor(max/3)
        tmp[1] = max % 3
        return tmp

    def randomize(self, r):
        for i in self.m:
            for k in i:
                k = k + math.random(0, r) - r/2
        for i in self.b:
            for k in i:
                k = k + math.random(0, r) - r/2

    def print(self):
        print(self.m)
        print(self.b)

    def copy(self):
        return MSGP(self.m, self.b)
```

```python
def deathBattle(a, b):
    b = board()
    turn = True
    for i in range(9):
        if(turn):
            move = a.move(b, 1)
        else:
            move = b.move(b, -1)
        x = move[0]
        y = move[1]
        val = b.move(x, y)
        if(val == -1):
            if(turn):
                return b
            else:
                return a
        if(b.checkWinTurn()):
            if(turn):
                return a
            else:
                return b
        turn = not turn
    return a
```

```python
population = 50 #MUST BE EVEN
generations = 40
hidden = 9
change = 2
moving = []
pop = []
    
for i in range(population):
    pop.append(MSGP(np.ones((hidden, 9)), np.ones((hidden, 9))))

for i in range(generations):
    for j in range(population):
        moving.append(deathBattle(pop[j], pop[population - j - 1]))
    pop = []
    for j in moving:
        pop.append(j.copy())
        pop.append(j.copy())
    for j in pop:
        j.randomize(change)
    
pop[0].print()


    
        
```
