```python
import numpy as np

class board:
    def __init__(self):
        self.arr = np.zeros((3,3))
        self.turn = True
    def print(self):
        for i in self.arr:
            print(i)
    def move(self, x, y):
        if(self.arr[x,y] == 0):
            if(self.turn):
                self.arr[x,y] = 1
            else:
                self.arr[x,y] = -1
            self.turn = not self.turn
            return 1
        else:
            return -1
    def winHelper(self, x, y, p):
        if(x > 2 or y > 2):
            return False
        if(self.arr[x,y] != p):
            return False
        
        checkX = True
        checkY = True
        for i in range(3):
            if(self.arr[i, y]):
                checkX = False
            if(self.arr[x, i]):
                checkY = False

        if(checkX):
            return True
        if(checkY):
            return True
        
        
        

        
        
b = board()

b.print()


```
