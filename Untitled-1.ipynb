```python
import numpy as np

class board:
    def __init__(self):
        self.arr = np.zeros((3,3))
        self.turn = True
    def print(self):
        for i in self.arr:
            print(i)
    def move(self, x, y):
        if(self.arr[x,y] == 0):
            if(self.turn):
                self.arr[x,y] = 1
            else:
                self.arr[x,y] = -1
            self.turn = not self.turn
            return 1
        else:
            return -1
        
b = board()
b.move(0,1)
b.print()


```
